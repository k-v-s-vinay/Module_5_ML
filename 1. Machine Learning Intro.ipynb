### Data Transformation Method (Reshaping)
1. To convert **non normal distribution** into **normal distribution**.
2. Various Transformations are:
*  * LogNormal
   * 

```python
# Data Distribution is verified using KDE plot
sns.kdeplot(X_train,x='fixed acidity',color = 'green',fill = True)
plt.title('Data Distribution of Fixed Acidity')
plt.show()
```

```python
# create QQ plot for fixed acidity column
stats.probplot(X_train['fixed acidity'],dist = 'norm',plot = plt)
plt.title('Verification of Normal Distribution - Fixed Acidity')
plt.show()
```

#### Observations:
1. here if all the blue color dots are fall on the exactly red line it will be a normal distribution
2. but here it is not a normal distribution

#### Power_Transform:
* box-cox
* yeo-johnson

```python
# apply box-cox transformation
from sklearn.preprocessing import PowerTransformer
pt = PowerTransformer(method = 'box-cox')
# apply transformation on X_train
X_train_bc = pt.fit_transform(X_train)
# apply transfrom X_test
X_test_bc = pt.transform(X_test)
```

```python
here box-cox cannot handle negative values so we can go with yeo-johnson
```

```python
# apply box-cox transformation
from sklearn.preprocessing import PowerTransformer
pt = PowerTransformer(method = 'yeo-johnson')
# apply transformation on X_train
X_train_bc = pt.fit_transform(X_train)
# apply transfrom X_test
X_test_bc = pt.transform(X_test)
```

```python
X_train_bc = pd.DataFrame(X_train_bc,columns = X_train.columns)
X_train_bc.head()
```

```python
X_test_bc = pd.DataFrame(X_test_bc,columns = X_test.columns)
X_test_bc.head()
```

```python
stats.probplot(X_train_bc['fixed acidity'],dist = 'norm',plot = plt)
plt.title('Verification of Normal Distribution - Fixed Acidity')
plt.show()
```

```python
stats.probplot(X_train_bc['volatile acidity'],dist = 'norm',plot = plt)
plt.title('After Transformation - Volatile acidity')
plt.show()
```

```python
sns.kdeplot(X_train_bc,x = 'fixed acidity',fill=True)
plt.title('Data Distribution After (yeo-Johnson)')
plt.show()
```
