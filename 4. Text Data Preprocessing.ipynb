#### Word Frequency Analysis:
1. counter is used to how times the word occurs

```python
from collections import Counter
# function to clean text and extract words
def get_words(text):
    txt = text.lower()
    txt = re.sub(r"[^a-zA-Z\s]","",txt)
    words = txt.split()
    return words
all_words = d['Text'].apply(get_words).sum()
# count the word frequencies
word_counts = Counter(all_words)
# display the most common words
word_counts.most_common(10)
```

#### Removing Stop words:

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
from nltk.corpus import stopwords
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))
stop_words
```

```python
t = "in computing. the term text processing refers to the theory and practical:"
# apply word tokenization
tokens = word_tokenize(t)
filtered_tokens = [word for word in tokens if word.lower() not in stop_words]
filtered_tokens
```

#### Stemming:
* it is a concept cut unwanted word
* it replaces the word by it's root word.
* dimensionality of the words can be reduced

```python
from nltk.stem import PorterStemmer
stemmer = PorterStemmer()
words = ['running','files','easily','fairly','doubtli']
stemmed_words = [stemmer.stem(word) for word in words]
stemmed_words
```

#### Lemmatization:
* the objective is same as stemming.
* but it replace the word in dictionary

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in words]
lemmatized_words
```
