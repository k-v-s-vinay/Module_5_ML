# **Types of Naive Bayes**

1. Gaussian Naive Bayes

2. Multinomial Naive Bayes

3. Bernoulli Naive Bayes

4. Complement Naive Bayes

5. Categorical Naive Bayes

6. Mixed Naive Bayes

# **1. Gaussian Naive Bayes**

- Assumes that the features follow a normal distribution.

- For each class, it estimates the mean (μ) and variance (σ²) of the features.

- Probability is calculated using the Gaussian probability density function

- Use case: When your features are continuous and roughly bell-shaped in distribution.


# **2. Multinomial Naive Bayes**

- Designed for count data, such as word frequencies in documents.

- Assumes features represent the number of times an event occurs.

- Often used with TF-IDF or Bag of Words models in NLP.

- Use case: Text classification, sentiment analysis, topic modeling.

# **3. Bernoulli Naive Bayes**

- Assumes binary features (e.g., word present or not).

- Each feature is modeled as a Bernoulli distribution (true/false).

- Suitable when you're only interested in presence/absence, not frequency.

- Use case: Spam filtering, document classification with binary features.

# **4. Complement Naive Bayes (CNB)**

- A variation of MultinomialNB, especially effective for imbalanced datasets.

- Often performs better in text classification tasks with skewed class distributions.

# **5. Categorical Naive Bayes**


- Categorical Naive Bayes is a variant of the Naive Bayes algorithm designed specifically for **categorical (nominal) features**—those that take on values from a fixed set of categories (like "red", "blue", "green").

- It assumes:
    - Features are **conditionally independent** given the class.

    - Each feature follows a **categorical distribution** (i.e., a multinomial distribution over a finite set of categories).


**How It Works:**

For each feature $x_i$ and class $y$, it estimates:

$$
P(x_i = c \mid y) = \frac{\text{count of } x_i = c \text{ in class } y + \alpha}{\text{total count of class } y + \alpha \cdot K}
$$

Where:
- $\alpha$ is the **smoothing parameter** (Laplace or Lidstone smoothing).
- $K$ is the number of categories for feature $x_i$.

Then, it uses Bayes Theorem to compute:

$$
P(y \mid \mathbf{x}) \propto P(y) \prod_{i=1}^{n} P(x_i \mid y)
$$


- **When to Use CategoricalNB:**

Use it when:
- Your features are **categorical**, not numerical or binary.
- You have **low-cardinality** features (e.g., "color" with values red, green, blue).
- You want a **fast, interpretable** model for classification.

# **6. Mixed Naive Bayes**

- **Mixed Naive Bayes** is a hybrid model that combines different Naive Bayes variants—like **GaussianNB**, **CategoricalNB**, and **BernoulliNB**—to handle datasets with **mixed feature types**:

- **Continuous features** → modeled using **Gaussian distribution**

- **Categorical features** → modeled using **Categorical (Multinomial) distribution**

- **Binary features** → modeled using **Bernoulli distribution**

- This allows you to apply the most appropriate statistical assumption for each feature type.


**Why Use MixedNB:**

- In real-world datasets, it's common to have a mix of:
    
    - Numerical values (e.g., age, income)
    
    - Categorical values (e.g., gender, occupation)
    
    - Binary flags (e.g., owns_car: yes/no)

- Using a single Naive Bayes variant (like GaussianNB) on all features can lead to poor performance.

- **MixedNB** solves this by applying the right distribution to each feature.

**How It Works:**

1. **Feature Segregation**:
   - Identify which features are **categorical** and which are **continuous**.

   - Optionally, identify **binary** features separately.

2. **Modeling**:

   - For each feature, it uses appropriate function
      
3. **Combine Probabilities**:

   - Multiply the probabilities from each feature type to compute the final class probabilities.

# **Smoothing Techniques Used In Naive Bayes**

- In Naive Bayes, **smoothing techniques** are essential for handling the **zero-frequency problem**.

- when a feature value does not appear in the training data for a given class, leading to a probability of zero and wiping out the entire prediction.

**1. Laplace Smoothing (Add-One Smoothing)**

- **What it does**: Adds 1 to every count to ensure no probability is zero.

- **Formula**:
  
  $$
  P(x_i \mid y) = \frac{\text{count}(x_i, y) + 1}{\text{count}(y) + k}
  $$
  where \( k \) is the number of possible values for \( x_i \).

- **Use case**: Widely used in text classification with MultinomialNB.

- **Pros**: Simple and effective.

- **Cons**: Can overly smooth when the vocabulary is large.


**2. Lidstone Smoothing (Add-α Smoothing)**

- **What it does**: Generalizes Laplace by adding a small constant $\alpha$  instead of 1.

- **Formula**:
  

  $$P(x_i \mid y) = \frac{\text{count}(x_i, y) + \alpha}{\text{count}(y) + \alpha \cdot k}$$

- **Typical α values**: Between 0.01 and 1.

- **Advantage**: More control over the degree of smoothing.



**3. Dirichlet Smoothing**

- **What it does**: A Bayesian approach that adjusts probabilities based on prior distributions.

- **Formula**:

  $$
  P(x_i \mid y) = \frac{\text{count}(x_i, y) + \mu \cdot P(x_i)}{\text{count}(y) + \mu}$$

  where $\mu$ is a hyperparameter and $P(x_i)$ is the background probability.

- **Use case**: Common in language modeling and information retrieval.


**4. Jelinek-Mercer Smoothing**

- **What it does**: Linearly interpolates between the observed probability and a background model.

- **Formula**:

  $$
  P(x_i \mid y) = (1 - \lambda) \cdot \frac{\text{count}(x_i, y)}{\text{count}(y)} + \lambda \cdot P(x_i)$$

- **Use case**: Often used in short text classification and IR tasks.


5. **Absolute Discounting**

- **What it does**: Subtracts a fixed discount d from each non-zero count and redistributes the leftover probability mass.

- **Formula**:

  $$
  P(x_i \mid y) = \frac{\max(\text{count}(x_i, y) - d, 0)}{\text{count}(y)} + \text{discount mass}$$

- **Use case**: Effective in sparse data scenarios.


**Why Smoothing Matters**

- Without smoothing:

   - A single unseen word can make the entire class probability zero.

   - The model becomes brittle and overfits to training data.

- With smoothing:

    - The model generalizes better to unseen data.

    - It avoids overconfidence in rare features.

# **Numerical Instability**

- Numerical instability refers to errors caused by limitations in floating-point precision when performing computations on very small or very large numbers.

- In Naive Bayes, this typically manifests as numerical underflow.

**Why It Happens in Naive Bayes**

- Naive Bayes computes the joint probability of a class and a feature vector by multiplying many small conditional probabilities:

$$P(y \mid \mathbf{x}) \propto P(y) \prod_{i=1}^{n} P(x_i \mid y) $$

- Each ( P(x_i \mid y) ) is a number between 0 and 1.

- Multiplying many such small numbers (e.g., ( $10^{-5} \times 10^{-6} \times \dots$ )) can result in values so small that they underflow to zero in floating-point arithmetic.

- Once a probability becomes zero, it dominates the entire computation, leading to incorrect predictions.


# **Logarithmic Transformation**

Instead of multiplying probabilities, take the logarithm and sum them:

$$\log P(y \mid \mathbf{x}) = \log P(y) + \sum_{i=1}^{n} \log P(x_i \mid y)$$

- Prevents underflow

- Turns multiplication into addition (which is numerically stable)
